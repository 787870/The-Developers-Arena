```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
import warnings
warnings.filterwarnings('ignore')

# 1. Load the transactional dataset
df = pd.read_csv(r'C:\Users\ajasl\OneDrive\Documents\Project\The Developers Arena\Month 5\data\supermarket_sales.csv')

# 2. Aggregate into daily sales time series
df['Date'] = pd.to_datetime(df['Date'])
daily_sales = df.groupby('Date')['Total'].sum().reset_index()
daily_sales = daily_sales.set_index('Date').sort_index()

print(f"Total days of data: {len(daily_sales)}")

# 3. Check for Stationarity (ADF Test)
print("\n--- Augmented Dickey-Fuller Test ---")
result = adfuller(daily_sales['Total'])
p_value = result[1]
print(f'ADF Statistic: {result[0]:.4f}')
print(f'p-value: {p_value:.4f}')

if p_value <= 0.05:
    print("Conclusion: The time series is stationary (Ready for modeling).")
else:
    print("Conclusion: The time series is non-stationary (Needs differencing).")
```
