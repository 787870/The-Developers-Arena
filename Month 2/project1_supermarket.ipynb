```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv(r'C:\Users\ajasl\OneDrive\Documents\Project\The Developers Arena\Month 2\supermarket_sales.csv')
```

```python
# Display basic dataset information
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
import os

# Create the visualizations folder if it doesn't already exist
os.makedirs('visualizations', exist_ok=True)
```

```python
# Group by Product Line to find total sales
product_sales = df.groupby('Product_Line')['Total'].sum().sort_values(ascending=False)
print("Total Sales by Product Line:")
print(product_sales)
```

```python
# Generate and save the first visualization
plt.figure(figsize=(10, 6))
sns.barplot(x=product_sales.values, y=product_sales.index, palette='Blues_r')
plt.title('Total Sales by Product Line - Supermarket Analysis')
plt.xlabel('Total Sales (₹)')
plt.ylabel('Product Line')
plt.tight_layout()
```

```python
# Save visualization
plt.savefig('visualizations/project1_product_sales.png')
plt.show()
```

```python
# Generate and save the second visualization (Payment Methods)
plt.figure(figsize=(8, 6))
payment_counts = df['Payment'].value_counts()
plt.pie(payment_counts, labels=payment_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Transaction Distribution by Payment Method')
plt.tight_layout()
```

```python
# Save visualization
plt.savefig('visualizations/project1_payment_methods.png')
plt.show()
```

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'])
```

```python
# Group by Date to get daily sales total
daily_sales = df.groupby('Date')['Total'].sum().reset_index()
```

```python
# Generate and save the third visualization (Line Trend)
plt.figure(figsize=(12, 6))
plt.plot(daily_sales['Date'], daily_sales['Total'], color='teal', linewidth=1.5)
plt.title('Daily Sales Trend Over Time - Supermarket Analysis')
plt.xlabel('Date')
plt.ylabel('Total Sales (₹)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
```

```python
# Save visualization
plt.savefig('visualizations/project1_sales_trend.png')
plt.show()
```
